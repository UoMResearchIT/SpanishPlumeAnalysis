```python
import xarray as xr
from wrf_widget import view_wrf_3d, view_wrf_2d
from pathlib import Path
```

```python
# Load a data file into an XArray DataSet object
data_dir = "/mnt/projects/wrf-storm-workshop/data/wrfout/arwen_ctrl/d01/"
filename = "wrfout_d01_2021-11-26_00:00:00"
ds = xr.open_dataset(
    str(Path(data_dir, filename)),
    engine="netcdf4"
)
```

```python
# Optional print all of the variables inside the DataSet
print(ds.data_vars)
```

```python
# Example plot of a 2D field
varname = "U10" # x-wind fields at 10m
# Load the variable into a 2D DataArray, with time dimension removed
da = ds[varname].isel(Time=0, drop=True)
view_wrf_2d(da)
```

```python
# Example plot of a 3D field 
varname = "U" # x-dimension 3D wind fields
# Load the variable into a 3D DataArray, with time dimension removed
da = ds[varname].isel(Time=0, drop=True)
view_wrf_3d(da, vmin=-10, vmax=10, cmap="bwr")
```
